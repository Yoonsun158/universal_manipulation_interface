```python
import numpy as np

duration = 10.0
sample_dt = 1 / 100
k = int(duration / sample_dt)
sample_t = np.linspace(0, duration, k)
value = np.sin(sample_t * duration / 1.5) * 0.5 + 0.5
width = value * 80
# print(sample_t)
# print(value)
print(width)
```

```python
# 示例：测量经过时间
import time
print(time.time())
for i in range(5):

    t_start = time.monotonic()

    print(t_start)

    # 做一些工作
    time.sleep(1.0)
    print(time.monotonic())
    elapsed = time.monotonic() - t_start
    print(f"{i}. Elapsed: {elapsed:.3f}s")
print(time.time())
```
